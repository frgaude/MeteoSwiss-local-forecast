# Boat

Forecast wind for the next 48 h in knots: hourly mean with its 10–90 % range, gusts and direction.

Data: MeteoSwiss local forecast (OGD), updated hourly. For the gridded ensemble over the lake, see the icon-ch1-eps-wind repo.

```python
import json
import localforecast as lf
import screens
```

```python
# Find point ids: search by town, station abbreviation or postal code
lf.find_point("Lutry")
```

```python
# ── Configuration ──────────────────────────────────────────────
POINT = (109500, 2)       # Lutry, postal code 1095, on the lake shore (392 m)
WIND_STATION = "CHVD19"   # WindSpots: Lutry - Port du Vieux-Stand (measured wind)
HOURS = 39       # chart: 39 h from the last 00:00 / 12:00
# ───────────────────────────────────────────────────────────────
```

```python
run, data = lf.fetch([POINT], lf.BOAT_PARAMS)
hourly, _ = data[POINT]
measured = lf.windspots_station(WIND_STATION)
view = lf.boat_view(run, hourly, lf.point_info(POINT), measured, hours=HOURS)
screens.render(screens.boat, view, colour=True)
```

```python
# As on the LilyGO T5 4.7" e-paper: 960×540, 16 grey levels
img = screens.render(screens.boat, view)
img
```

```python
# The payload a display would receive
print(f"{len(json.dumps(view, separators=(',', ':')))} bytes as JSON, "
      f"{len(screens.epd_buffer(img)):,} bytes as a 4-bit frame buffer")
print(json.dumps(view, indent=1)[:1500])
```
